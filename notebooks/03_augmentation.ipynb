```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)

NOTEBOOK_DIR = Path.cwd()
if NOTEBOOK_DIR.name != "notebooks":
    NOTEBOOK_DIR = NOTEBOOK_DIR / "notebooks"
PREPARATION_NOTEBOOK = NOTEBOOK_DIR / "02_data_preparation.ipynb"

if not PREPARATION_NOTEBOOK.exists():
    raise FileNotFoundError(f"Preparation notebook was not found: {PREPARATION_NOTEBOOK}")

# Reuse the prepared X_train and y_train variables from notebook 02.
get_ipython().run_line_magic("run", str(PREPARATION_NOTEBOOK))

if "X_train" not in globals() or "y_train" not in globals():
    raise RuntimeError("Notebook 02 did not create X_train and y_train.")

print("Prepared training images:", X_train.shape)
print("Prepared training labels:", y_train.shape)
print("Validation and test data are not loaded into the augmentation pipeline.")
```

```python
train_datagen = ImageDataGenerator(
    rotation_range=10,
    width_shift_range=0.08,
    height_shift_range=0.08,
    zoom_range=0.10,
    horizontal_flip=False,
    fill_mode="nearest",
)

train_generator = train_datagen.flow(
    X_train,
    y_train,
    batch_size=32,
    shuffle=True,
    seed=RANDOM_STATE,
)

print("Training augmentation is configured.")
print("Rotation: +/-10 degrees")
print("Width and height shifts: +/-8%")
print("Zoom: +/-10%")
print("Horizontal flip: disabled")
```

```python
sample_count = 6
sample_indices = np.arange(sample_count)
original_images = X_train[sample_indices]
original_labels = y_train[sample_indices]

augmented_images = np.stack([
    train_datagen.random_transform(image, seed=RANDOM_STATE + index)
    for index, image in enumerate(original_images)
])
augmented_images = np.clip(augmented_images, 0.0, 1.0)

figure, axes = plt.subplots(2, sample_count, figsize=(15, 5))
for index, (original, augmented, label) in enumerate(zip(original_images, augmented_images, original_labels)):
    axes[0, index].imshow(original)
    axes[0, index].set_title(f"Original\nClass {label}")
    axes[1, index].imshow(augmented)
    axes[1, index].set_title(f"Augmented\nClass {label}")
    axes[0, index].axis("off")
    axes[1, index].axis("off")

axes[0, 0].set_ylabel("Original", fontsize=12)
axes[1, 0].set_ylabel("Augmented", fontsize=12)
figure.suptitle("Examples of Training-Only Augmentation")
figure.tight_layout()
plt.show()
```
